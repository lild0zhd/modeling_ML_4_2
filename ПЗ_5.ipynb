```python
import math
import numpy as np
```

```python
# f1(x) = (x - 2)^2 + 1
def f1(x):
    return (x - 2)**2 + 1

def df1(x):
    return 2 * (x - 2)

def d2f1(x):
    return 2


# f2(x) = (x + 1)^4 + 0.5 * (x + 1)^2
def f2(x):
    return (x + 1)**4 + 0.5 * (x + 1)**2

def df2(x):
    return 4 * (x + 1)**3 + (x + 1)

def d2f2(x):
    return 12 * (x + 1)**2 + 1


# f3(x) = x^4 - 3x^3 + 2
def f3(x):
    return x**4 - 3*x**3 + 2

def df3(x):
    return 4*x**3 - 9*x**2

def d2f3(x):
    return 12*x**2 - 18*x


# f4(x) = |x - 1|
def f4(x):
    return abs(x - 1)


# f5(x) = (x - 3)^2 + sin(5x)
def f5(x):
    return (x - 3)**2 + math.sin(5*x)

def df5(x):
    return 2*(x - 3) + 5*math.cos(5*x)

def d2f5(x):
    return 2 - 25*math.sin(5*x)


# Небольшая проверка
print("f1(2) =", f1(2))
print("f2(-1) =", f2(-1))
print("f3(0) =", f3(0))
print("f4(1) =", f4(1))
print("f5(3) =", f5(3))
```

```python
def dichotomy_min(f, a, b, eps=1e-6, delta=1e-7, max_iter=1000):
    history = []       # история шагов
    f_calls = 0        # считаем число вычислений функции

    for i in range(max_iter):
        mid = (a + b) / 2

        # две точки рядом с серединой
        x1 = mid - delta
        x2 = mid + delta

        f_x1 = f(x1)
        f_x2 = f(x2)
        f_calls += 2

        history.append((i + 1, a, b, x1, x2, f_x1, f_x2))

        # если интервал уже достаточно маленький, завершаем
        if (b - a) / 2 < eps:
            x_star = (a + b) / 2
            return x_star, f(x_star), f_calls + 1, history

        # оставляем ту часть интервала, где функция меньше
        if f_x1 <= f_x2:
            b = x2
        else:
            a = x1

    x_star = (a + b) / 2
    return x_star, f(x_star), f_calls + 1, history


# Пример запуска
x_star, f_star, f_calls, hist = dichotomy_min(f1, -5, 5, eps=1e-5, delta=1e-6)

print("Метод дихотомии")
print("x* =", x_star)
print("f(x*) =", f_star)
print("Число вычислений f =", f_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

Здесь задаются тестовые функции из задания, а также производные там, где они нужны.
Это основа для всех последующих методов.
В конце добавлена простая проверка, что функции работают.

```python
def golden_section_min(f, a, b, eps=1e-6, max_iter=1000):
    phi = (1 + math.sqrt(5)) / 2   # золотое отношение
    resphi = 2 - phi               # коэффициент для пересчета точек

    # начальные внутренние точки
    x1 = a + resphi * (b - a)
    x2 = b - resphi * (b - a)

    f1_val = f(x1)
    f2_val = f(x2)
    f_calls = 2
    history = []

    for i in range(max_iter):
        history.append((i + 1, a, b, x1, x2, f1_val, f2_val))

        # если интервал стал маленьким, завершаем
        if abs(b - a) < eps:
            x_star = (a + b) / 2
            return x_star, f(x_star), f_calls + 1, history

        # переиспользуем одну из старых точек, чтобы экономить вычисления
        if f1_val <= f2_val:
            b = x2
            x2 = x1
            f2_val = f1_val
            x1 = a + resphi * (b - a)
            f1_val = f(x1)
            f_calls += 1
        else:
            a = x1
            x1 = x2
            f1_val = f2_val
            x2 = b - resphi * (b - a)
            f2_val = f(x2)
            f_calls += 1

    x_star = (a + b) / 2
    return x_star, f(x_star), f_calls + 1, history


# Пример запуска
x_star, f_star, f_calls, hist = golden_section_min(f1, -5, 5, eps=1e-5)

print("Метод золотого сечения")
print("x* =", x_star)
print("f(x*) =", f_star)
print("Число вычислений f =", f_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

Метод золотого сечения тоже сужает интервал, но делает это экономнее по числу вычислений функции.
На каждом шаге одна из старых точек сохраняется, поэтому не нужно заново считать всё с нуля.
Из-за этого метод обычно выгоднее обычной дихотомии.

```python
def parabolic_interpolation_min(f, a, b, eps=1e-6, max_iter=100):
    # берем три начальные точки
    x1 = a
    x3 = b
    x2 = (a + b) / 2

    f1_val = f(x1)
    f2_val = f(x2)
    f3_val = f(x3)
    f_calls = 3
    history = []

    for i in range(max_iter):
        # коэффициенты формулы для минимума параболы по трем точкам
        numerator = ((x2 - x1)**2) * (f2_val - f3_val) - ((x2 - x3)**2) * (f2_val - f1_val)
        denominator = 2 * ((x2 - x1) * (f2_val - f3_val) - (x2 - x3) * (f2_val - f1_val))

        # если знаменатель слишком мал, делаем подстраховочный шаг в центр
        if abs(denominator) < 1e-14:
            x_hat = (a + b) / 2
        else:
            x_hat = x2 - numerator / denominator

        # если новая точка вышла за интервал, берем подстраховочный шаг
        if not (a <= x_hat <= b):
            x_hat = (a + b) / 2

        f_hat = f(x_hat)
        f_calls += 1
        history.append((i + 1, a, b, x1, x2, x3, x_hat, f_hat))

        # если шаг маленький, завершаем
        if abs(x_hat - x2) < eps:
            return x_hat, f_hat, f_calls, history

        # обновляем тройку точек так, чтобы минимум оставался внутри
        points = [(x1, f1_val), (x2, f2_val), (x3, f3_val), (x_hat, f_hat)]
        points = sorted(points, key=lambda t: t[0])

        # выбираем три соседние точки вокруг лучшей
        best_index = min(range(len(points)), key=lambda j: points[j][1])

        if best_index == 0:
            new_points = points[:3]
        elif best_index == len(points) - 1:
            new_points = points[-3:]
        else:
            new_points = points[best_index - 1:best_index + 2]

        (x1, f1_val), (x2, f2_val), (x3, f3_val) = new_points
        a, b = x1, x3

    return x2, f2_val, f_calls, history


# Пример запуска
x_star, f_star, f_calls, hist = parabolic_interpolation_min(f2, -3, 3, eps=1e-5)

print("Метод параболической интерполяции")
print("x* =", x_star)
print("f(x*) =", f_star)
print("Число вычислений f =", f_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

Этот метод строит параболу по трём точкам и берёт минимум этой параболы как новое приближение.
Если шаг получается неудачным, используется безопасная замена.
Метод может работать быстрее золотого сечения, но он менее устойчив.

```python
def hybrid_brent_like_min(f, a, b, eps=1e-6, max_iter=100):
    # начальная лучшая точка
    x = (a + b) / 2
    w = x
    v = x

    fx = f(x)
    fw = fx
    fv = fx
    f_calls = 1
    history = []

    for i in range(max_iter):
        use_parabola = False
        x_try = None

        # пытаемся сделать параболический шаг, если точки различны
        if len(set([x, w, v])) == 3:
            numerator = ((x - w)**2) * (fx - fv) - ((x - v)**2) * (fx - fw)
            denominator = 2 * ((x - w) * (fx - fv) - (x - v) * (fx - fw))

            if abs(denominator) > 1e-14:
                x_try = x - numerator / denominator

                # используем шаг только если он попал внутрь интервала
                if a <= x_try <= b:
                    use_parabola = True

        # если параболический шаг не подошел, берем шаг по золотому сечению
        if not use_parabola:
            if x < (a + b) / 2:
                x_try = x + 0.381966 * (b - x)
            else:
                x_try = x - 0.381966 * (x - a)

        f_try = f(x_try)
        f_calls += 1
        history.append((i + 1, a, b, x, x_try, fx, f_try))

        # обновляем интервал и лучшие точки
        if f_try < fx:
            if x_try < x:
                b = x
            else:
                a = x
            v, fv = w, fw
            w, fw = x, fx
            x, fx = x_try, f_try
        else:
            if x_try < x:
                a = x_try
            else:
                b = x_try

            if f_try < fw or w == x:
                v, fv = w, fw
                w, fw = x_try, f_try
            elif f_try < fv or v == x or v == w:
                v, fv = x_try, f_try

        if abs(b - a) < eps:
            return x, fx, f_calls, history

    return x, fx, f_calls, history


# Пример запуска
x_star, f_star, f_calls, hist = hybrid_brent_like_min(f3, -1, 4, eps=1e-5)

print("Гибридный метод")
print("x* =", x_star)
print("f(x*) =", f_star)
print("Число вычислений f =", f_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

Это упрощённый гибридный метод: сначала он пытается сделать быстрый параболический шаг.
Если такой шаг неудачен, используется более надёжный шаг по схеме золотого сечения.
Идея в том, чтобы совместить скорость и устойчивость.

```python
def bisection_derivative(df, a, b, eps=1e-6, max_iter=1000):
    if df(a) * df(b) > 0:
        raise ValueError("На концах интервала производная должна иметь разные знаки.")

    history = []
    df_calls = 2

    for i in range(max_iter):
        c = (a + b) / 2
        df_c = df(c)
        df_calls += 1

        history.append((i + 1, a, b, c, df_c))

        # если производная близка к нулю, нашли стационарную точку
        if abs(df_c) < eps or abs(b - a) < eps:
            return c, df_calls, history

        if df(a) * df_c < 0:
            b = c
        else:
            a = c
        df_calls += 1   # потому что снова считаем df(a) на следующей проверке

    return c, df_calls, history


# Пример запуска
x_star, df_calls, hist = bisection_derivative(df1, -5, 5, eps=1e-5)

print("Бисекция для f'(x)=0")
print("x* =", x_star)
print("f1(x*) =", f1(x_star))
print("Число вычислений df =", df_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

Если функция гладкая, минимум можно искать как решение уравнения f'(x)=0.
Метод бисекции делает это надёжно, если производная на концах интервала имеет разные знаки.
Этот подход хорош, когда производная известна явно.

```python
def newton_min(df, d2f, x0, eps_x=1e-6, eps_grad=1e-6, max_iter=100):
    x = x0
    history = []
    df_calls = 0
    d2f_calls = 0

    for i in range(max_iter):
        grad = df(x)       # первая производная
        hess = d2f(x)      # вторая производная
        df_calls += 1
        d2f_calls += 1

        # защита от деления на маленькое число
        if abs(hess) < 1e-14:
            print("Остановка: вторая производная слишком мала.")
            return x, df_calls, d2f_calls, history

        # формула метода Ньютона
        x_new = x - grad / hess

        history.append((i + 1, x, grad, hess, x_new, abs(x_new - x)))

        # критерии остановки
        if abs(x_new - x) < eps_x or abs(grad) < eps_grad:
            return x_new, df_calls, d2f_calls, history

        x = x_new

    return x, df_calls, d2f_calls, history


# Пример запуска
x_star, df_calls, d2f_calls, hist = newton_min(df1, d2f1, x0=0, eps_x=1e-5)

print("Метод Ньютона")
print("x* =", x_star)
print("f1(x*) =", f1(x_star))
print("Число вычислений df =", df_calls)
print("Число вычислений d2f =", d2f_calls)
print("Все шаги:")
for row in hist:
    print(row)
```

Метод Ньютона для минимума использует первую и вторую производные.
На каждом шаге новое приближение вычисляется через отношение f'(x) к f''(x).
Метод часто очень быстрый, но чувствителен к выбору начальной точки.

```python
def gradient_descent_fixed(df, f, x0, alpha=0.1, eps=1e-6, max_iter=1000):
    x = x0
    history = []
    f_calls = 0
    df_calls = 0

    for i in range(max_iter):
        grad = df(x)   # градиент в текущей точке
        df_calls += 1

        # шаг против градиента
        x_new = x - alpha * grad

        f_old = f(x)
        f_new = f(x_new)
        f_calls += 2

        history.append((i + 1, x, grad, x_new, f_old, f_new))

        # если изменение стало маленьким, завершаем
        if abs(x_new - x) < eps:
            return x_new, f_calls, df_calls, history

        x = x_new

    return x, f_calls, df_calls, history


# Пример запуска
x_star, f_calls, df_calls, hist = gradient_descent_fixed(df5, f5, x0=0, alpha=0.05, eps=1e-5)

print("Градиентный метод с фиксированным шагом")
print("x* =", x_star)
print("f5(x*) =", f5(x_star))
print("Число вычислений f =", f_calls)
print("Число вычислений df =", df_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```

В градиентном методе новое значение ищется шагом против производной.
Если шаг выбран удачно, функция убывает.
Но при неудачном шаге возможны перелёты и колебания.

```python
def gradient_descent_backtracking(df, f, x0, alpha0=1.0, beta=0.5, c=1e-4, eps=1e-6, max_iter=1000):
    x = x0
    history = []
    f_calls = 0
    df_calls = 0

    for i in range(max_iter):
        grad = df(x)   # производная в текущей точке
        df_calls += 1

        alpha = alpha0   # начинаем с большого шага
        f_x = f(x)
        f_calls += 1

        # уменьшаем шаг, пока условие Армихо не выполнено
        while f(x - alpha * grad) > f_x - c * alpha * (grad ** 2):
            alpha = beta * alpha
            f_calls += 1

        # выполняем шаг после подбора alpha
        x_new = x - alpha * grad

        history.append((i + 1, x, grad, alpha, x_new, f_x, f(x_new)))
        f_calls += 1

        if abs(x_new - x) < eps:
            return x_new, f_calls, df_calls, history

        x = x_new

    return x, f_calls, df_calls, history


# Пример запуска
x_star, f_calls, df_calls, hist = gradient_descent_backtracking(df5, f5, x0=0, alpha0=1.0, beta=0.5, c=1e-4, eps=1e-5)

print("Градиентный метод с backtracking")
print("x* =", x_star)
print("f5(x*) =", f5(x_star))
print("Число вычислений f =", f_calls)
print("Число вычислений df =", df_calls)
print("Первые 5 шагов:")
for row in hist[:5]:
    print(row)
```
